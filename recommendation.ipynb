```python
# recommendation 

def get_recommendations(hours, attendance, tutoring, sleep):

    recommendations = []

    if hours < 15:
        recommendations.append("Increase your study hours.")

    if attendance < 75:
        recommendations.append("Improve your attendance.")

    if tutoring < 2:
        recommendations.append("Consider additional tutoring sessions.")

    if sleep < 6:
        recommendations.append("Try to get enough sleep.")

    if len(recommendations) == 0:
        recommendations.append("Good job! Keep maintaining your current habits.")

    return recommendations
```

```python
recommendations = get_recommendations(10, 65, 0, 5)

for r in recommendations:
    print("→", r)
```
